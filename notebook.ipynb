```python
#           remove entries from Dataframe:
df1 = df1.drop(39) # will delete whatevers specified / returned in the curly brackets
#   (has no output as this is an assignment operation)
#   this will delete row 39, so no more "Anna Ivanova"
#--try: `df1.tail()` to see that the last 5 rows end at only 38
```

```python
#           clean data, either within df or by returning new df
df1.dropna()# drop anything containing N/A / null values: if deletes individ element, or entire row idk
# returns new df, so would need to be used in an assignment, `=`, i guess
df1.fillna({"OrderID":0}, inplace=True)# for any null OrderID value, replace it with a '0'
#   `inplace` means this command modifies the current dataframe instead of returning a new one
df1.rename(columns={"OrderID":"Order ID"}, inplace=True)# renames whats in blue (ie columns)
#   modifying instead of returning a new df
#   df1.info() to check the renaming has worked
```

```python
#           analyse data
df1["Country"].value_counts()# use to see how many values of specified category (ie from each country)
#   output counts down from largest to smallest numbers
#df1.groupby("Country")["Price"].sum()# sort by multiple filters
```

```python
#--df1.groupby("Country")["Price"].sum()# sort by multiple filters, ie add up orders from each country
#   with country still being first, apparently that means the category name and is alphabetical
#--df1.sort_values("Price", ascending=False)# sorts price descending, ie from largest down to smallest
#    so what about sorting the total country prices as well?
df1.sort_values(df1.groupby("Country")["Price"].sum(), ascending=False)
```
